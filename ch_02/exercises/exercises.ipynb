1. Find the 95th percentile of earthquake magnitude in Japan using the mb magnitude type.

```python
df[df['parsed_place'] == 'Japan']['mag'].quantile(0.95)
```

2. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.

```python
indo = df[(df['parsed_place'] == 'Indonesia')]
indo_tsu = df[(df['parsed_place'] == 'Indonesia') & (df['tsunami'] == 1)]

percent_indo_tsu = indo_tsu.parsed_place.count() / indo.parsed_place.count() 
round(percent_indo_tsu, 2)   
```

4. Add a column indicating whether the earthquake happened in a country or US state that is on the Ring of Fire. Use Alaska, Antarctica (look for Antarctic), Bolivia, California, Canada, Chile, Costa Rica, Ecuador, Fiji, Guatemala, Indonesia, Japan, Kermadec Islands, Mexico (be careful not to select New Mexico), New Zealand, Peru, Philippines, Russia, Taiwan, Tonga, and Washington.

```python
us_mask = df['parsed_place'].isin(['Alaska', 'California', 'Washington'])
ring_mask = df['parsed_place'].isin(['Antarctica', 'Bolivia', 'Canada', 'Chile', 'Costa Rica', 'Ecuador', 'Fiji', 'Guatemala', 'Indonesia', 'Japan', 'Kermadec Islands', 'Mexico', 'New Zealand', 'Peru', 'Philippines', 'Russia', 'Taiwan', 'Tonga'])
df = df.assign(
    region=np.where(
        us_mask,
        'US',
        np.where(ring_mask, 'Ring of Fire', 'Other')
    )
)
df.head(2)
```

5. Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them.

```python
earthquakes_by_region = df.groupby('region').size().reset_index(name='count')
earthquakes_in_us = earthquakes_by_region[earthquakes_by_region['region'] == 'US']['count'].iloc[0]
earthquakes_elsewhere = earthquakes_by_region[earthquakes_by_region['region'] != 'US']['count'].sum()
earthquakes_in_us, earthquakes_elsewhere
```

6. Find the tsunami count along the Ring of Fire.

```python
df[df['region'] == 'Ring of Fire']['tsunami'].sum()
```
