## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random

random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
```

## Exercise 5: Calculating statistics and verifying
### mean

```python
mean = sum(salaries) / len(salaries)
mean
```

### median

```python
median = sorted(salaries)[len(salaries)//2 if len(salaries) % 2 != 0 else (len(salaries)//2 - 1 + len(salaries)//2) // 2]
median
```

### mode

```python
from collections import Counter
mode = Counter(salaries).most_common(1)[0][0]
mode
```

### sample variance
Remember to use Bessel's correction.

```python
sample_var = (sum((x - mean) ** 2 for x in salaries) / (len(salaries) - 1)) 
sample_var
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sample_sd = (sum((x - mean) ** 2 for x in salaries) / (len(salaries) - 1)) ** 0.5
sample_sd
```

## Exercise 6: Calculating more statistics
### range

```python
range = max(salaries) - min(salaries)
range
```

### coefficient of variation
Make sure to use the sample standard deviation.

```python
coef_var = sample_sd // mean
coef_var
```

### interquartile range

```python
import numpy as np
q1, q3 = np.quantile(salaries, [0.25, 0.75], method='linear') 
inter_quartile_range = q3 - q1
```

### quartile coefficent of dispersion

```python
quartile_cof_dispersion = inter_quartile_range / q1 + q3
quartile_cof_dispersion
```

## Exercise 7: Scaling data
### min-max scaling

```python
import numpy as np
mm_normalized_salaries = (salaries - np.min(salaries)) / range
mm_normalized_salaries
```

### standardizing

```python
normalized_salaries = (salaries - np.array(mean)) / sample_sd
normalized_salaries
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
import numpy as np
cov_matrix = np.cov(mm_normalized_salaries, normalized_salaries)
cov_matrix
```

### Pearson correlation coefficient ($\rho$)

```python
mm_sd, n_sd = np.std(mm_normalized_salaries, ddof=1), np.std(normalized_salaries, ddof=1)
corr = cov_matrix / (mm_sd * n_sd)
corr
```
